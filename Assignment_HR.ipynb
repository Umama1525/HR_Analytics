```python
from google.colab import drive
drive.mount('/content/drive')
import seaborn as sns
import pandas as pd
import matplotlib.pyplot as plt
```

```python
# Load dataset
df = pd.read_csv('/content/drive/MyDrive/HR_Analytics.csv')

# Display basic information
df.head()
```

```python
# Filter numerical columns
numerical_df = df.select_dtypes(include=['number'])

# Display statistics
for col in numerical_df.columns:
    mean = numerical_df[col].mean()
    median = numerical_df[col].median()
    var = numerical_df[col].var()
    std = numerical_df[col].std()

    print(f"Statistics for Column: {col}")
    print(f"Mean: {mean:.2f}")
    print(f"Median: {median:.2f}")
    print(f"Variance: {var:.2f}")
    print(f"Standard Deviation: {std:.2f}")
    print("-" * 40)
```

```python
# Bar plots 1
plt.figure(figsize=(8, 5))
df['Department'].value_counts().plot(kind='bar', title='Employees per Department')
plt.xlabel('Department')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# Bar plots 2
plt.figure(figsize=(8, 5))
df['EducationField'].value_counts().plot(kind='bar', title='Count of Employees by Education Field')
plt.xlabel('Education Field')
plt.ylabel('Count')
plt.tight_layout()
plt.show()
```

```python
# line graphs 1
plt.figure(figsize=(8, 5))
df.groupby('Age')['MonthlyIncome'].mean().plot(kind='line', title='Average Monthly Income by Age')
plt.xlabel('Age')
plt.ylabel('Average Monthly Income')
plt.tight_layout()
plt.show()
```

```python
plt.figure(figsize=(8, 5))
df.groupby('YearsAtCompany')['JobSatisfaction'].mean().plot(kind='line', title='Job Satisfaction by Years at Company')
plt.xlabel('Years at Company')
plt.ylabel('Average Job Satisfaction')
plt.tight_layout()
plt.show()
```

```python
# Pie Chart 1

plt.figure(figsize=(6, 6))
df['Gender'].value_counts().plot(kind='pie', autopct='%1.1f%%', title='Gender Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Pie Chart 2

plt.figure(figsize=(6, 6))
df['Attrition'].value_counts().plot(kind='pie', autopct='%1.1f%%', title='Attrition Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# Compute the correlation matrix (only numerical data)
correlation_matrix = df.corr(numeric_only=True)

# Plot the heatmap
plt.figure(figsize=(12, 8))
sns.heatmap(correlation_matrix, annot=True, cmap='coolwarm', fmt=".2f")
plt.title('Correlation Heatmap of Numerical Features')
plt.tight_layout()
plt.show()
```
